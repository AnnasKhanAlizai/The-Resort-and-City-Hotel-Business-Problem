```python
fig = sns.countplot(x = 'hotel', hue = 'is_canceled', data = df, palette = 'Accent')
plt.title('Resort and City Hotel Reservation status', size=15)
plt.xlabel('Hotels')
plt.ylabel('No. of Reservations')
plt.legend(['Not Cancelled', 'Cancelled'])
plt.show()
```

```python
#Resort Hotel percentage of reservation

resort_hotel = df[df['hotel'] == 'Resort Hotel']
resort_hotel['is_canceled'].value_counts(normalize = True)
percentage = ((resort_hotel['is_canceled'].value_counts(normalize = True)*100))
percentage = round(percentage, 1)
percentage
```

```python
#City Hotel percentage of reservation

city_hotel = df[df['hotel'] == 'City Hotel']
city_hotel['is_canceled'].value_counts(normalize = True)
percentage = ((city_hotel['is_canceled'].value_counts(normalize = True)*100))
percentage = round(percentage, 1)
percentage
```

```python

resort_hotel = resort_hotel.groupby('reservation_status_date')[['adr']].mean()
city_hotel = city_hotel.groupby('reservation_status_date')[['adr']].mean()

# for self understanding

#print('Resort Hotel', resort_hotel)
#print('================================')
#print('City Hotel', city_hotel)
```

```python
plt.figure(figsize = (20,10))
plt.title('Avg. daily rate in City and Resort Hotel', fontsize = 30)
plt.scatter(resort_hotel.index, resort_hotel['adr'], label = 'Resort Hotel', color = 'red')

plt.scatter(city_hotel.index, city_hotel['adr'], label = 'City Hotel', color = 'green')

plt.legend(fontsize = 21)

plt.show()
```

```python


plt.figure(figsize = (20,10))
plt.title('Avg. daily rate in City and Resort Hotel', fontsize = 30)
plt.plot(resort_hotel.index, resort_hotel['adr'], label = 'Resort Hotel', color = 'red')

plt.plot(city_hotel.index, city_hotel['adr'], label = 'City Hotel', color = 'green')



plt.legend(loc = 'best')

plt.show()
```

```python
# from fig we can see on an average, rates of City Hotel is low, but in some days span Resort Hotel decreases its rates below the average rates
# of City Hotel.

# now we see which month has much reservations and which has low.
```

```python
# create a month column
df['month'] = df['reservation_status_date'].dt.month
plt.figure(figsize = (20,8))
a = sns.countplot(x = 'month', hue = 'is_canceled', data = df, palette = 'BuGn')

plt.grid()
plt.title("Reservation Status by Months", size = 25)
plt.ylabel("No. of Reservations")
plt.legend(['Not Cancelled', 'cancelled'],fontsize = 12)
plt.show()
```

```python
# Average daily Revenue

plt.figure(figsize=(20, 8))

sns.barplot(x='month', y='adr', data=df[df['is_canceled'] == 1].groupby('month')[['adr']].sum().reset_index(), color = 'green')
plt.title('ADR (Avg. Daily Revenue) by Month', fontsize =25)
plt.grid()
plt.show()

```

```python
cancelled_data = df[df['is_canceled'] == 1]
top_12_country = cancelled_data['country'].value_counts()[:12]
plt.figure(figsize = (20,8))

plt.pie(top_12_country, autopct = '%.2f', labels = top_12_country.index)
plt.title('Top 12 Countries Cancelation %age')
plt.show()
```

### Below we check customers booking trend. via Online, Walk-in, or by Agent.

```python
Trend_plot = df['market_segment'].value_counts()
Trend_plot
```

```python
# %age of Ways of reservations

Trend_plot = df['market_segment'].value_counts(normalize = True)
Trend_plot = round((Trend_plot)*(100))
Trend_plot
```

```python
# %age of which Way have more cancelleations of reservation.

Cancellations_trend = cancelled_data['market_segment'].value_counts(normalize = True)
Cancellations_trend = round((Cancellations_trend)*(100))
Cancellations_trend 
```

```python
figure = plt.figure(figsize = (5,15))
fig = px.line(df['market_segment'].value_counts(), markers = True)

fig.update_layout(title="Customers Booking Trend")
fig.show()
```

```python
# now we check rate of cancelation basis on prices


cancelled_df_adr= cancelled_data.groupby(['reservation_status_date'])[['adr']].mean()
cancelled_df_adr.reset_index( inplace=True)
cancelled_df_adr.sort_values('reservation_status_date', inplace = True)

not_cancelled_data = df[df['is_canceled'] == 0]
not_cancelled_df_adr= not_cancelled_data.groupby(['reservation_status_date'])[['adr']].mean()
not_cancelled_df_adr.reset_index( inplace=True)
not_cancelled_df_adr.sort_values('reservation_status_date', inplace = True)

plt.figure(figsize=(20, 6))
plt.title("Average Daily Rate by Reservation Status")
plt.plot(not_cancelled_df_adr["reservation_status_date"], not_cancelled_df_adr['adr'], label='Not Cancelled')
plt.plot(cancelled_df_adr['reservation_status_date'], cancelled_df_adr['adr'], label='Cancelled')
plt.legend()

plt.show()
```

```python
# Clearly price rates are higher on an average and also cause of cancelation.
```

```python
cancelled_df_adr = cancelled_df_adr[(cancelled_df_adr["reservation_status_date"]>'2016') & (cancelled_df_adr["reservation_status_date"]<'2017-09')]
not_cancelled_df_adr= not_cancelled_df_adr[(not_cancelled_df_adr['reservation_status_date']>'2016') & (not_cancelled_df_adr['reservation_status_date']<'2017-09')]
```

```python
plt.figure(figsize=(30, 6))
plt.title("Daily Average Rate of Reservation Status by price", fontsize = 30)
plt.plot(not_cancelled_df_adr["reservation_status_date"], not_cancelled_df_adr['adr'], label='Not Cancelled')
plt.plot(cancelled_df_adr['reservation_status_date'], cancelled_df_adr['adr'], label='Cancelled')
plt.legend(fontsize = 21)

plt.show()
```
